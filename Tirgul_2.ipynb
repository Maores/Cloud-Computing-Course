```python
path = "/content/drive/MyDrive/s1.json"

# Load students list
with open(path, "r", encoding="utf-8") as f:
    students = json.load(f)

# Validating the list exists
if isinstance(students, dict):
    students = [students]

print("Loaded", len(students), "students")
```

```python
#Using widgets on the form
# UI widgets for the form (read-only except favorite)
import ipywidgets as w
from IPython.display import display, clear_output

# (2) Build the list of names (fallback to empty string if missing)
names = [f"{s.get('name','')} {s.get('surname','')}" for s in students]

dd = w.Dropdown(options=names, description="Student")
name_tb = w.Text(description="First", disabled=True)
surname_tb = w.Text(description="Last", disabled=True)
mail_tb = w.Text(description="Email", disabled=True)
courses_tb = w.Text(description="Courses", disabled=True)  # comma-separated view
link_tb = w.Text(description="Link", disabled=True)

favorite_tb = w.Text(description="Favorite program")  # editable
save_btn = w.Button(description="Update JSON", button_style="success")
out = w.Output()

display(dd, name_tb, surname_tb, mail_tb, courses_tb, link_tb, favorite_tb, save_btn, out)

#(3) When user selects a student, populate the fields from `students[idx]`
def on_select(change):
    if dd.index is None or dd.index < 0:
        return
    s = students[dd.index]
    name_tb.value = s.get("name","")
    surname_tb.value = s.get("surname","")
    mail_tb.value = s.get("mail","")
    # Show courses as a comma-separated string
    crs = s.get("courses", [])
    courses_tb.value = ", ".join(crs if isinstance(crs, list) else [str(crs)])
    link_tb.value = s.get("nice_link", s.get("link",""))
    favorite_tb.value = s.get("favorite_program","")  # might be missing initially

# (4) Wire the handler and initialize once
dd.observe(on_select, names="value")
on_select(None)

# (5) Save the updated favorite_program back to the JSON file
def on_save(_):
    with out:
        clear_output()
        try:
            idx = dd.index
            if idx is None or idx < 0:
                print("No student selected.")
                return
            # Update only favorite_program
            students[idx]["favorite_program"] = favorite_tb.value

            with open(path, "w", encoding="utf-8") as f:
                json.dump(students, f, ensure_ascii=False, indent=2)

            print(f" Saved. Updated favorite_program for: {names[idx]}")
        except Exception as e:
            print(" Error while saving:", e)

save_btn.on_click(on_save)

```
